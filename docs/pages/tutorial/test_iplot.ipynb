# Smoke tests for `iplot`

Converted to a pytest module by `pixi run test` (via
`jupyter nbconvert --to script`). Each `test_*` function below is picked up
by pytest; keep assertions inside functions named `test_*` so they survive
the conversion as real, individually reported tests.

```python
import matplotlib
matplotlib.use("Agg")  # headless: no display available under nbconvert/pytest

import pandas as pd
import ipywidgets as widgets

from iplot_widget import iplot, iplot_settings
import iplot_widget.iplot_widget as _impl
```

```python
data = pd.DataFrame({
    "x": [1, 2, 3, 4, 5, 6],
    "y": [2, 4, 1, 3, 5, 2],
    "species": ["a", "a", "b", "b", "c", "c"],
})
```

```python
def test_iplot_returns_output_widget():
    result = iplot(data, x="x", y="y", hue="species")
    assert isinstance(result, widgets.Output)
```

```python
def test_iplot_settings_updates_options():
    original = _impl.OPTIONS["max_figure_width"]
    try:
        iplot_settings(max_figure_width=12)
        assert _impl.OPTIONS["max_figure_width"] == 12
    finally:
        iplot_settings(max_figure_width=original)
```

```python
def test_ax2fig_covers_common_plot_kinds():
    for kind in ["scatter", "line", "hist", "box", "bar"]:
        assert kind in _impl.AX2FIG
```
